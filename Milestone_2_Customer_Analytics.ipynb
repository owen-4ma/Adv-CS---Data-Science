### In this assignment I have been given a data set customer_train.csv a subset of a larger data set for all customers, in the aims of storing the data more efficiently. the requirements I have been given are:
- Columns containing categories with only two factors must be stored as Booleans (bool).

- Columns containing integers only must be stored as 32-bit integers - (int32).

- Columns containing floats must be stored as 16-bit floats (float16).

- Columns containing nominal categorical data must be stored as the category data type.

# **Converting Categories**
Now that we've determined what to do with the data, we can move to actually carrying that out. First convert the series relevant_experience and job_change to booleans.

```python
#since job_change is already numbered with 0s and 1s, astype can be used to convert the column into a bool.
ds_jobs_transformed['job_change'] = ds_jobs_transformed['job_change'].astype('bool')
#check to confirm the change
print(ds_jobs_transformed['job_change'].dtype)
```

```python
#relevant_experience is challenging since it has values as text, using the
#map method and passing a dictionary through with the key as old value
# and the value as the new bool is a simple way to get this done
ds_jobs_transformed['relevant_experience'] = ds_jobs_transformed['relevant_experience'].map({'Has relevant experience': True, "No relevant experience": False})

#the column still is not stored as a boolean so call the astype method again

ds_jobs_transformed['relevant_experience'] = ds_jobs_transformed['relevant_experience'].astype('bool')

#final check to ensure that it has been safe

print(ds_jobs_transformed['relevant_experience'].dtype)
```

Next converting columns that are integers into int32s instead of of int64s

```python
#simple as type method with int32 as parameter
ds_jobs_transformed['training_hours'] = ds_jobs_transformed['training_hours'].astype('int32')
print(ds_jobs_transformed['training_hours'].dtype)
```

next is converting city_development_index to float16

```python
ds_jobs_transformed['city_development_index'] = ds_jobs_transformed['city_development_index'].astype('float16')
print(ds_jobs_transformed['city_development_index'].dtype)
```

Nominal data is trickier to work with as it has no natural order, but they only need to be sorted as categories. whereas ordinal data has to be set. We know what categories to set because of the value_counts analysis done earlier

```python
#using Series function with data type specified as category to change
ds_jobs_transformed['gender'] = pd.Series(ds_jobs_transformed['gender'], dtype = 'category')
#check if data type is correct
print(ds_jobs_transformed['gender'].dtype)
#check if categories are in correctly
print(ds_jobs_transformed['gender'].head())

#repeat with all nominal data series

#enrolled_university
ds_jobs_transformed['enrolled_university'] = pd.Series(ds_jobs_transformed['enrolled_university'], dtype = 'category')

print(ds_jobs_transformed['enrolled_university'].dtype)
print(ds_jobs_transformed['enrolled_university'].head())

#major_discipline
ds_jobs_transformed['major_discipline'] = pd.Series(ds_jobs_transformed['major_discipline'], dtype = 'category')

print(ds_jobs_transformed['major_discipline'].dtype)
print(ds_jobs_transformed['major_discipline'].head())

#company_type
ds_jobs_transformed['company_type'] = pd.Series(ds_jobs_transformed['company_type'], dtype = 'category')

print(ds_jobs_transformed['company_type'].dtype)
print(ds_jobs_transformed['company_type'].head())


#student_id
ds_jobs_transformed['student_id'] = pd.Series(ds_jobs_transformed['student_id'], dtype = 'category')

print(ds_jobs_transformed['student_id'].dtype)
print(ds_jobs_transformed['student_id'].head())

#city
ds_jobs_transformed['city'] = pd.Series(ds_jobs_transformed['city'], dtype = 'category')

print(ds_jobs_transformed['city'].dtype)
print(ds_jobs_transformed['city'].head())
```

# **Converting to Ordinal Data**
ordinal data has a set hierarchy because of the code and requires more programming

```python
#use pd.categorical and pass a list of the desired categories in order (lowest to highest), and make sure the parameter ordered is set to True
ds_jobs_transformed['education_level'] = pd.Categorical(ds_jobs_transformed['education_level'], ['Primary School','High School','Graduate','Masters',' PhD'], ordered = True)
#checking for data type and categories
print(ds_jobs_transformed['education_level'].head())

```

This one was quite simple because it only contained five categories, however some can be more complicated in which case it is better to use a list

```python
#manually creating a list comprehension this way is faster than iterating through the series since we already know the
experience_order = ['<1'] + [str(i) for i in range(1, 21)] + ['>20']

ds_jobs_transformed['experience'] = pd.Categorical(ds_jobs_transformed['experience'], experience_order, ordered = True)
#checking for data type and categories
print(ds_jobs_transformed['experience'].value_counts(dropna = False))


last_job_order = ['<1'] + [str(i) for i in range(1, 5)] + ['>4']

ds_jobs_transformed['last_new_job'] = pd.Categorical(ds_jobs_transformed['last_new_job'], last_job_order, ordered = True)
#checking for data type and categories
print(ds_jobs_transformed['last_new_job'].value_counts(dropna = False))

#manually created list
company_size_order = ['<10','10-49','50-99','100-499','500-999','1000-4999','5000-9999','10000+']

#insert that into the pd.Categorical method, the NaN values were interferring with the conversion and causing all values to be returned as NaN, converting to str and replacing with the real numpy NaN fixed this issue
ds_jobs_transformed['company_size'] = pd.Categorical(ds_jobs_transformed['company_size'].astype(str).replace('nan',np.nan),categories=company_size_order, ordered = True)
#checking for data type and categories
print(ds_jobs_transformed['company_size'].value_counts(dropna = False))




```

# **Filtering Through Dataframe**
The last requirement of this dataset was focusing on a specific group, the assignment requried the final dataframe to contain only people who have at least 10 or more years of experience at companies with at least 1000 employees.

```python
#filter based off of the already created categories, allowing for easy comparison with these operators
ds_jobs_transformed = ds_jobs_transformed[(ds_jobs_transformed['experience'] >= '10') &(ds_jobs_transformed['company_size'] >= '1000-4999')]
#check to make sure the filtering worked
print(ds_jobs_transformed['experience'].value_counts())
print(ds_jobs_transformed['company_size'].value_counts())

#check for datatype and memory usage
print(ds_jobs_transformed.info())

#comparison of memory usage with original dataset, 1.3Mb difference
print(ds_jobs.info())
```
