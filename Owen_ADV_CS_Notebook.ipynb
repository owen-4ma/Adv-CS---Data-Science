# **Joining data**
- The default wait to join data is using the merge() method in pandas.
- call it on one data frame and another data frame as an arguement and it will combine the two
- on parameter specifies which specific column to jon on
- how parameter is how the data will be joined, default being inner where both keys are combined. A join can be thought of like an venn diagram, with an inner join consisting of overlapping rows.
    - outer is all of the rows that do not overlap
    - left contains all the values from the left dataframe as well as the overlapping ones with the right
    - right is essentially the same thing but without the
- suffix parameter takes in a tuple of two suffixes that will distinguish two identical columns from each other when they are not being used in the join
- left_on and right_on arguments allow you to specify which column from the right and which column from the left to merge during the join.
- Self Join is when you join a dataframe with itself. This is useful in situations involving hierarchal relationships, sequential relationships,
**Relationships between dataframes**
- one to one: every row in the left table is related to one and only one row in the right table.
- one to many: every row in the left table is related to one or more rows in the right table.
- in the case of one to many, it means there are certain values in rows from the first dataframe that multiple values from the second dataframe contain. Pandas takes care of the difference so no difference in syntax is necessary

**Merging Multiple Dataframes**
- if you need to merge multiple columns pass a list through the on parameter
- to merge twice, just call merge in the function twice, not forgetting to include the backslash so as to continue the function as one.

  - step 1: Inner join the two tables
